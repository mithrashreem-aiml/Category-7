```python
# Map Coloring using Backtracking
# With step-by-step iteration

colors = ["Red", "Green", "Blue"]

graph = {
    "A": ["B", "C"],
    "B": ["A", "C", "D"],
    "C": ["A", "B", "D"],
    "D": ["B", "C"]
}

step = 0

def is_safe(node, color, assignment):
    for neighbor in graph[node]:
        if neighbor in assignment and assignment[neighbor] == color:
            return False
    return True


def map_coloring(assignment):
    global step

    # If all nodes are assigned
    if len(assignment) == len(graph):
        print("\nAll nodes are colored!")
        return assignment

    # Select first unassigned node
    node = next(n for n in graph if n not in assignment)

    print("\nSelecting node:", node)

    for color in colors:
        step += 1

        print("\nStep", step)
        print("Trying", node, "=", color)

        if is_safe(node, color, assignment):
            print("Result: SAFE")

            assignment[node] = color

            print("Current assignment:", assignment)

            result = map_coloring(assignment)

            if result:
                return result

            # Backtracking
            print("Backtracking from", node)
            del assignment[node]

        else:
            print("Result: CONFLICT")


# Start the algorithm
solution = map_coloring({})

print("\n-----------------------------")
print("Final Map Coloring Solution")
print("-----------------------------")
print(solution)
```

```python
#Local search in CSP (SUDOKU)
import random

grid = [
    [5, 3, 0, 0, 7, 0, 0, 0, 0],
    [6, 0, 0, 1, 9, 5, 0, 0, 0],
    [0, 9, 8, 0, 0, 0, 0, 6, 0],
    [8, 0, 0, 0, 6, 0, 0, 0, 3],
    [4, 0, 0, 8, 0, 3, 0, 0, 1],
    [7, 0, 0, 0, 2, 0, 0, 0, 6],
    [0, 6, 0, 0, 0, 0, 2, 8, 0],
    [0, 0, 0, 4, 1, 9, 0, 0, 5],
    [0, 0, 0, 0, 8, 0, 0, 7, 9]
]

original = [row[:] for row in grid]


def conflicts(grid):
    count = 0

    # Rows
    for row in grid:
        count += 9 - len(set(row))

    # Columns
    for col in range(9):
        values = [grid[row][col] for row in range(9)]
        count += 9 - len(set(values))

    # 3x3 boxes
    for r in range(0, 9, 3):
        for c in range(0, 9, 3):
            values = []
            for i in range(r, r + 3):
                for j in range(c, c + 3):
                    values.append(grid[i][j])
            count += 9 - len(set(values))

    return count


# Fill empty cells randomly
for i in range(9):
    missing = [n for n in range(1, 10) if n not in grid[i]]
    random.shuffle(missing)

    for j in range(9):
        if grid[i][j] == 0:
            grid[i][j] = missing.pop()


# Local Search
for _ in range(100000):

    if conflicts(grid) == 0:
        break

    r = random.randint(0, 8)
    c = random.randint(0, 8)

    if original[r][c] != 0:
        continue

    r2 = random.randint(0, 8)
    c2 = random.randint(0, 8)

    if original[r2][c2] != 0:
        continue

    old = conflicts(grid)

    grid[r][c], grid[r2][c2] = grid[r2][c2], grid[r][c]

    new = conflicts(grid)

    if new > old:
        grid[r][c], grid[r2][c2] = grid[r2][c2], grid[r][c]


print("Sudoku Solution:")
for row in grid:
    print(*row)
```
